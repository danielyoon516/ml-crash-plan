```python
"""
2. Randomly initialize w_hat, b_hat. Compute predictions and MSE using only tensor ops, with no nn and no F.mse_loss.
	3.	Derive the gradients of MSE with respect to w_hat and b_hat on paper, then implement them as tensor ops.
	4.	Write a gradient descent loop and run it until w_hat is close to the true w. Print the loss every N steps.
"""
```

```python
import torch

torch.manual_seed(0)
X = torch.randn(100, 3)
w = torch.randn(3,1)
b = torch.randn(1,1)
noise = 0.1 * torch.randn(100,1)
y = X @ w + b + noise
assert y.shape == (100,1)
```

```python
w_hat = torch.randn(3,1)
b_hat = torch.randn(1,1)
y_hat = X @ w_hat + b_hat
mse = ((y_hat - y)**2).mean()
assert y_hat.shape == (100,1)
assert (y_hat - y).shape == (100,1)
print(mse)
```

```python
N = X.shape[0]
grad_y_hat = (2/N)* (y_hat - y)
assert grad_y_hat.shape == (100,1)
grad_w = X.T @ grad_y_hat
assert grad_w.shape == (3,1)
grad_b = grad_y_hat.sum(dim=0, keepdim=True)
assert grad_b.shape == (1,1)
```

```python
w_hat = torch.randn(3, 1)
b_hat = torch.randn(1, 1)
lr = 0.1
N = X.shape[0]

for step in range(200):
    y_hat = X @ w_hat + b_hat
    assert y_hat.shape == (100,1)
    mse = ((y_hat - y)**2).mean()
    assert (y_hat - y).shape == (100,1)
    grad_y_hat = (2/N)* (y_hat - y)
    assert grad_y_hat.shape == (100,1)
    grad_w = X.T @ grad_y_hat
    assert grad_w.shape == (3,1)
    grad_b = grad_y_hat.sum(dim=0, keepdim=True)
    assert grad_b.shape == (1,1)
    w_hat = w_hat - lr * grad_w
    b_hat = b_hat - lr * grad_b
    if step % 20 == 0:
        print(step, mse.item())

print(torch.cat([w_hat, w], dim = 1))
print(b_hat, b)
print(mse.item())
```
