```python
np.random.seed(0)
X = np.random.rand(100) * 10
y = 3 * X + 5 + np.random.randn(100)

import matplotlib.pyplot as plt
plt.scatter(X, y)
plt.show()
```

```python
w, b = 3, 5
preds = (X*w + b)
print(preds.shape)
```

```python
def cost(w, b, X, y):
    m = len(X)
    preds = (X*w + b)
    errors = preds - y
    return np.sum((errors ** 2) / (2*m))


print(cost(3, 5, X, y))
print(cost(0, 0, X, y))
```

```python
ws = np.linspace(0, 6, 50)
costs = [cost(w, 5, X, y) for w in ws]

plt.plot(ws, costs)
plt.xlabel("w")
plt.ylabel("cost")
plt.show()
```
